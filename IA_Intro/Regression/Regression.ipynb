```python
X=np.array(df.YearsExperience)
```

```python
y=np.array(df.Salary)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
reg=LinearRegression().fit(X_train.reshape(-1, 1),y_train.reshape(-1, 1))
```

```python
from sklearn.metrics import mean_squared_error
y_pred_train=reg.predict(X_train.reshape(-1, 1))
y_pred_test=reg.predict(X_test.reshape(-1, 1))
```

```python
mse_train=mean_squared_error(y_pred_train,y_train)
print('mean_squared_error y_pred_train is ',mse_train)
mse_test=mean_squared_error(y_pred_test,y_test)
print('mean_squared_error y_pred_train is ',mse_test)
```

```python
print('R square train ',reg.score(X_train.reshape(-1, 1),y_train))
print('R square train ',reg.score(X_test.reshape(-1, 1),y_test))
```

```python
#adjusted R-square score
from sklearn.metrics import r2_score
p=1#number of observations /number of regressors
n=len(X_test)# sample size
print(1-(1-r2_score(y_test, y_pred_test))*((n/n-p-1)))
```

```python
import matplotlib.pyplot as plt 
plt.figure(1)
plt.plot(X_train, y_train, 'yo', X_train, y_pred_train, '--k')
plt.xlabel("Nombre de voisins (k)") 
plt.ylabel("Précision") 
#plt.grid(True) 
plt.show()
```

```python
plt.figure(2)
plt.plot(X_test, y_test, 'yo', X_test, y_pred_test, '--k')
plt.xlabel("Nombre de voisins (k)") 
plt.ylabel("Précision") 
#plt.grid(True) 
plt.show()
```
