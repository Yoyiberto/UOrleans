```python
 from sklearn.ensemble import RandomForestClassifier
 from sklearn.datasets import load_iris
 from sklearn.model_selection import train_test_split
 from sklearn.metrics import accuracy_score
 # Load data
 X, y = load_iris(return_X_y=True)
 X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3)
 # Train model
 clf = RandomForestClassifier(n_estimators=100, random_state=42)
 clf.fit(X_train, y_train)
 # Predict
 y_pred = clf.predict(X_test)
 # Evaluate
 print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
import numpy as np 
import matplotlib.pyplot as plt 
# Function to minimize: f(x) = (x - 3)^2 
def f(x): 
    return (x - 3) ** 2 
 
# Gradient of the function: df/dx = 2(x - 3) 
def gradient(x): 
    return 2 * (x - 3) 
 
# Gradient descent 
def gradient_descent(starting_point, learning_rate, num_iterations): 
    x = starting_point 
    history = [x]  # To store x values at each iteration 
    for _ in range(num_iterations): 
        grad = gradient(x) 
        x = x - learning_rate * grad 
        history.append(x) 
    return x, history 
 
# Parameters 
starting_point = 10  # Initial starting point 
learning_rate = 0.1  # Learning rate 
num_iterations = 25  # Number of iterations 
 
# Apply gradient descent 
final_x, history = gradient_descent(starting_point, learning_rate, num_iterations) 
 
# Plotting the results 
x_values = np.linspace(0, 10, 400) 
y_values = f(x_values) 
 
plt.plot(x_values, y_values, label='f(x) = (x - 3)^2') 
plt.scatter(history, [f(x) for x in history], color='red', label='Iterations') 
plt.title(f'Gradient Descent (learning rate = {learning_rate})') 
plt.xlabel('x') 
plt.ylabel('f(x)') 
plt.legend() 
plt.grid(True) 
plt.show() 
print(f"Final value of x: {final_x}") 
print(f"Final value of f(x): {f(final_x)}") 
```

```python
# Visualizing how f(x) decreases with each iteration 
errors = [f(x) for x in history] 
plt.plot(errors) 
plt.title('Decrease of f(x) at Each Iteration') 
plt.xlabel('Iterations') 
plt.ylabel('f(x)') 
plt.grid(True) 
plt.show()
```
